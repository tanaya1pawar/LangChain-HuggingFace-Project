```python
!pip install -q --upgrade numpy langchain-huggingface
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline
from langchain_huggingface import HuggingFacePipeline
```

```python
model_id = "gpt2"
model = AutoModelForCausalLM.from_pretrained(model_id)
tokenizer = AutoTokenizer.from_pretrained(model_id)
```

```python
pipe = pipeline("text-generation", model=model, tokenizer=tokenizer, max_new_tokens=50)
local_llm = HuggingFacePipeline(pipeline=pipe)

response = local_llm.invoke("What is machine learning?")
print(response)
```

```python
!pip install -q langchain langchain-core
import sys

# Force both common local installation paths into sys.path
for path in ['/usr/local/lib/python3.13/site-packages', '/usr/local/lib/python3.13/dist-packages']:
    if path not in sys.path:
        sys.path.insert(0, path)

from langchain_core.prompts import PromptTemplate

template = "Question: {question}\nAnswer: Let's think step by step."
prompt = PromptTemplate(template=template, input_variables=["question"])
```

```python
chain = prompt | local_llm
```

```python
response = chain.invoke({"question": "What is machine learning?"})
print(response)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

model_name = "sentence-transformers/all-mpnet-base-v2"
model_kwargs = {"device": "cpu"}
encode_kwargs = {"normalize_embeddings": False}

embeddings = HuggingFaceEmbeddings(
    model_name=model_name,
    model_kwargs=model_kwargs,
    encode_kwargs=encode_kwargs
)
```

```python
text_list = ["Hello world!", "LangChain with Hugging Face embeddings."]
doc_embeddings = embeddings.embed_documents(text_list)

query_embedding = embeddings.embed_query("What is machine learning?")

print(f"Number of documents embedded: {len(doc_embeddings)}")
print(f"Dimension of query embedding: {len(query_embedding)}")
```

```python
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2")
```

```python
texts = [
    "Generative AI models combined with LangChain allow you to build powerful custom applications.",
    "FAISS is an efficient similarity search and clustering dense vectors library.",
    "Vector databases store text embeddings for fast retrieval."
]

db = FAISS.from_texts(texts, embeddings)
```

```python
query = "What is FAISS used for?"
results = db.similarity_search(query, k=1)

print(results[0].page_content)
```

```python
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2")

loaded_db = FAISS.load_local("faiss_index", embeddings, allow_dangerous_deserialization=True)
```

```python
query = "Tell me about vector databases."
results = loaded_db.similarity_search(query, k=2)

for i, doc in enumerate(results):
    print(f"Result {i+1}: {doc.page_content}")
```

```python
query = "What is FAISS?"
results_with_scores = loaded_db.similarity_search_with_score(query, k=2)

for doc, score in results_with_scores:
    print(f"Score: {score:.4f} | Content: {doc.page_content}")
```

```python
retriever = loaded_db.as_retriever(search_kwargs={"k": 2})
```

```python
from langchain_core.prompts import PromptTemplate

template = """Answer the question based only on the following context:
{context}

Question: {question}
"""
prompt = PromptTemplate.from_template(template)
```

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | local_llm  # Replace with your local_llm or llm instance
    | StrOutputParser()
)
```

```python
response = rag_chain.invoke("What is FAISS used for?")
print(response)
```
